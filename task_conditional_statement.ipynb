## Car Rental System Case Study - conditional statement case study


```python
# Task 1: Define Rental Rates Based on Car Type

car_type = input("Enter the type of car (Economy, Standard, Luxury): ").lower()

if car_type == 'economy':
    rental_rate = 50
elif car_type == 'standard':
    rental_rate = 75
elif car_type == 'luxury':
    rental_rate = 100
else:
    print("Invalid car type entered. Please choose from Economy, Standard, or Luxury.")

print(f"Car Type: {car_type}")
```

```python
# Task 2: Calculate Rental Duration Fee
rental_duration = int(input("Enter the rental duration in days: "))

if rental_duration <=0 :
    print("Invalid rental duration. Please enter a positive number of days.")
else:
    total_rental_fee = rental_rate * rental_duration
    print(f"The total rental fee for a {car_type.capitalize()} car for {rental_duration} days is: ${total_rental_fee}")
```

```python
# Task 3: Apply Membership Discount

membership_type = input("Enter your membership type (None, Standard, Premium): ").lower()

if membership_type == 'premium':
    discount_rate = 0.15
elif membership_type == 'standard':
    discount_rate = 0.10
elif membership_type == 'none':
    discount_rate = 0.0
    print("No discount applied.")
else:
    print("Invalid membership type entered. Please choose from None, Standard, or Premium.")

discount_amount = total_rental_fee * discount_rate
print(f"The discount amount for a {membership_type.capitalize()} membership is: ${discount_amount}")

total_rental_fee_after_discount = total_rental_fee * (1 - discount_rate)
print(f"The total rental fee after applying the {membership_type.capitalize()} membership discount is: ${total_rental_fee_after_discount}")
```

```python
# Task 4: Add Optional Services
insurance_ = input("Do you want to add insurance for $20/day? (yes/no): ").lower()
if insurance_ == 'yes':
    insurance_fee = 20 * rental_duration
else:
    insurance_fee = 0

print(f"The total insurance fee is: ${insurance_fee}")

gps_ = input("Do you want to add GPS for $5/day? (yes/no): ").lower()
if gps_ == 'yes':
    gps_fee = 5 * rental_duration
else:
    gps_fee = 0

print(f"The total GPS fee is: ${gps_fee}")

# Total Optional Services Fee
total_optional_services_fee = insurance_fee + gps_fee
print(f"The total optional services fee is: ${total_optional_services_fee}")
```

```python
# Task 5: Calculate Late Return Penalty

late_return_days = int(input("Enter the number of late return days: "))

if late_return_days < 0:
    print("Invalid input. Please enter a non-negative number of days.")
else:
    penalty = 30 * late_return_days
    print(f"The late return penalty is: ${penalty}")
```

```python
total_car_fee = 0

if rental_duration > 0 and membership_type in ['premium', 'standard', 'none'] and car_type in ['economy', 'standard', 'luxury'] and late_return_days >= 0 and insurance_ in ['yes', 'no'] and gps_ in ['yes', 'no']:
    total_car_fee = (total_rental_fee_after_discount + total_optional_services_fee + penalty)
    print(f"The total rental fee for a {car_type.capitalize()} car for {rental_duration} days with {membership_type.capitalize()} membership is: ${total_car_fee} with a late return penalty of ${penalty} and optional services fee of ${total_optional_services_fee}.")
```
