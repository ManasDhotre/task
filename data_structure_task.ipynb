```python
# 9. Add 10 to an integer and then subtract 5 from it. Print the result.
add_and_sub = (num1 + 10) - 5
print(add_and_sub)
```

```python
# 10. Calculate simple interest given P, R, T (No loop needed).
p = 30000 # Principal -> 30000
r = 8     # Rate of interest -> 8 
t = 2     # Tenure -> 2 years

simple_interest= (p * r * t) / 100

print("Simple Interest is : ", simple_interest)
```

## SET

```python
# 51. Create a set with 3 numbers and print it.
num_set = {2, 5, 8}
num_set2 = {4, 6, 5}
print(num_set, type(num_set))
```

```python
# 52. Add a new element to the set.
num_set.add(30)
print(num_set)
```

```python
# 53. Remove an element from the set.
num_set.remove(2)
print(num_set)
```

```python
# 54. Check if an element exists in the set.
print(5 in num_set)
```

```python
# 55. Find the union of two sets.
union_set = num_set.union(num_set2)
print(num_set ,'\n',num_set2)
print(union_set)
```

```python
# 56. Find the intersection of two sets.
inter_set = num_set.intersection(num_set2)
print(inter_set)
```

```python
# 57. Find the difference of two sets.
diff_set = num_set - num_set2
print(diff_set)
```

```python
# 58. Copy a set to a new variable.
num_set_copy = num_set.copy()
num_set_copy.add('Hello')

print(num_set)
print(num_set_copy)
```

```python
# 59. Check if a set is a subset of another.
print(num_set2.issubset(num_set))
```

```python
# 60. Clear all elements from a set.
num_set.clear()
print(num_set)
```
